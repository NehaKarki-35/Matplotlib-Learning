```python
import matplotlib.pyplot as plt
tracking_place=["Bara Bhangal","Pin Bhaba Pass","Annapurna Base trek","Roopkund","Bali Pass Trek","Rupin","Everest Base Camp Trek","Friendship Peak","Kuari Pass", "Nag Tibba Trek"]
sales =  [52000, 18000, 24800, 12300, 3400, 75000, 30500, 21000,45000,32000]

plt.figure(figsize=(10,6))
plt.xticks(rotation=45, ha="right")
plt.plot(tracking_place, sales, marker="o",color="#FF5733", linewidth=2, markersize=8, linestyle='--')
plt.title("Monthly Sales Growth per Trekking Place")
plt.xlabel("Tracking Place")
plt.ylabel("Sales (INR")
plt.grid(alpha=0.3)

for i, value in enumerate(sales):
    plt.text(i, value+1500, f"{value:,}", ha='center', fontsize=9, fontweight='bold')

plt.show()
```
